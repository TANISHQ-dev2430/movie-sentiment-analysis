```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent=['the glass of milk','the glass of juice','the cup of tea','I am going to the market','iam a good boy','iam a good developer','understand the meaning of words','your videos are good','i love to watch your videos']
```

```python
sent
```

```python
voc_size=10000
```

```python
#one hot representation
onehot_repr=[one_hot(words,voc_size)for words in sent]
onehot_repr
```

```python
#word embedding 
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_length=8
embedded_docs=pad_sequences(onehot_repr,padding='pre',maxlen=sent_length) ##pre-->>front padding and post-->>end padding
embedded_docs
```

```python
#feature representation
dim=10 
```

```python
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
